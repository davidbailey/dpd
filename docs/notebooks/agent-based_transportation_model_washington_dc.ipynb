# Agent-based Transportaion Model (Washington DC)

```python
from pyrosm import OSM, get_data

region = "district-of-columbia"
fp = get_data(region)
osm = OSM(fp)
osm.keep_node_info = True
network = osm.get_network(
    "all",
    extra_attributes=[
        "lanes:forward",
        "lanes:backward",
        "cycleway:left",
        "cycleway:right",
    ],
    nodes=True,
)
```
